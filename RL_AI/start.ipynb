```python
# 5. 학습 실행
import importlib
import os
import sys
from pathlib import Path
from shutil import which
import torch

home = Path.home()
rl_ai_root = home / "RL_AI"
if rl_ai_root.exists() and str(home) not in sys.path:
    sys.path.insert(0, str(home))

# 모듈 리로드 및 캐시 삭제
for module_name in list(sys.modules):
    if module_name == "RL_AI" or module_name.startswith("RL_AI."):
        del sys.modules[module_name]
importlib.invalidate_caches()

from RL_AI.training.experiment import run_checkpoint_training_experiment
from RL_AI.agents.agents import SeaEngineRLAgent, SeaEngineGreedyAgent, SeaEngineRandomAgent

TOTAL_EPISODES  = 10000
EVAL_INTERVAL   = 600
EVAL_MATCHES    = 100
UPDATE_INTERVAL = 16
SEED            = 7
card_data_path  = str(rl_ai_root / "SeaEngine/Cards.csv")

agent = SeaEngineRLAgent(device="cuda" if torch.cuda.is_available() else "cpu", seed=SEED)
opponent_pool = [
    SeaEngineRandomAgent(seed=SEED + 10),
    SeaEngineGreedyAgent(seed=SEED + 20),
]

print(f"Starting Experiment: {TOTAL_EPISODES} eps, update every {UPDATE_INTERVAL}")

checkpoint = await run_checkpoint_training_experiment(
    agent=agent,
    eval_matches=EVAL_MATCHES,
    total_train_episodes=TOTAL_EPISODES,
    eval_interval=EVAL_INTERVAL,
    update_interval=UPDATE_INTERVAL,
    train_opponent_pool=opponent_pool,
    card_data_path=card_data_path,
    seed=SEED,
)

print("\n=== Checkpoint Training Complete ===")
print(f"Summary: {checkpoint['summary_report_path']}")
print(f"Total time: {checkpoint['total_time_sec']:.0f}s")

# 모델 저장 (models 폴더 하위로 변경)
models_dir = rl_ai_root / "models"
models_dir.mkdir(parents=True, exist_ok=True)
agent.save(models_dir / "my_best_model.pt")
print(f"[Model Final Saved] => {models_dir / 'my_best_model.pt'}")
```

```python
# 6. 8방향 정밀 매트릭스 평가
from RL_AI.simulation.evaluator import SeaEngineComprehensiveEvaluator, print_matrix_report

print("=== 8-Way Matrix Evaluation (800 Matches) Start ===")
evaluator = SeaEngineComprehensiveEvaluator(agent=agent, card_data_path=card_data_path, num_envs=8)
opponents = [SeaEngineRandomAgent(), SeaEngineGreedyAgent()]

matrix_results = await evaluator.run_full_matrix(opponents=opponents, num_matches_per_case=100)
print_matrix_report(matrix_results)
```

```python
# 7. [신규 지표] 심층 분석 자동화 (Deep Analytics - 16방향 지원)
import pandas as pd
from IPython.display import display

def analyze_matrix_results(results):
    rows = []
    for opp, cases in results.items():
        for case, data in cases.items():
            # 예: "P1_Orange_vs_Charlotte" -> ["P1", "Orange", "vs", "Charlotte"]
            parts = case.split("_")
            side = parts[0]
            rl_deck = parts[1]
            opp_deck = parts[3]
            
            rows.append({
                "Opponent": opp,
                "Side": side,
                "RL_Deck": rl_deck,
                "Opp_Deck": opp_deck,
                "Type": "Mirror" if rl_deck == opp_deck else "Counter",
                "WinRate": data["win_rate"],
                "AvgSteps": data["avg_steps"]
            })
    
    df = pd.DataFrame(rows)
    
    print("\n=== [분석 1] 덱 상성별 승률 매트릭스 (Counter Match) ===")
    counter_df = df[df["Type"] == "Counter"]
    display(counter_df.pivot_table(values="WinRate", index=["Opponent", "RL_Deck"], columns="Side"))
    
    print("\n=== [분석 2] 동일 덱 대전 승률 (Mirror Match - 지능의 척도) ===")
    mirror_df = df[df["Type"] == "Mirror"]
    display(mirror_df.pivot_table(values="WinRate", index=["Opponent", "RL_Deck"], columns="Side"))
    
    print("\n=== [수석 엔지니어의 정밀 진단] ===")
    
    # Mirror Match 승률이 50%를 넘는지 확인 (실력 검증)
    mirror_avg = mirror_df["WinRate"].mean()
    if mirror_avg > 55:
        print(f"[실력] 훌륭합니다! 동일 덱 조건에서 상대보다 우월한 지능({mirror_avg:.1f}%)을 보여주고 있습니다.")
    elif mirror_avg < 45:
        print(f"[실력] 경고: 동일 덱 조건에서 상대 알고리즘에게 밀리고 있습니다({mirror_avg:.1f}%). 전략 학습이 더 필요합니다.")
    else:
        print(f"[실력] 현재 지능이 상대(Greedy/Random)와 대등한 수준({mirror_avg:.1f}%)입니다.")

    # 덱 간 불균형 체크
    orange_wr = df[df["RL_Deck"]=="Orange"]["WinRate"].mean()
    clear_wr = df[df["RL_Deck"]=="Charlotte"]["WinRate"].mean()
    print(f"[밸런스] 평균 승률: 귤 덱 {orange_wr:.1f}% vs 샤를 덱 {clear_wr:.1f}%")
    
    avg_steps = df["AvgSteps"].mean()
    print(f"[운영] 평균 교전 시간: {avg_steps:.1f} steps")

if 'matrix_results' in locals():
    analyze_matrix_results(matrix_results)
else:
    print("매트릭스 평가 결과가 없습니다. 6번 셀을 먼저 실행해 주세요.")
```
